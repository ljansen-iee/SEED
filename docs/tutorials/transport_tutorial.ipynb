# Transport Module — Step-by-step Tutorial

This notebook walks through the transport sector pipeline interactively.

```python
import sys, pathlib

repo_root = pathlib.Path().resolve().parents[1]
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

print("Python:", sys.version)
print("Repo root:", repo_root)
```

## 2 — Configuration overview

The transport module is configured via `transport/trans_config.csv`.

```python
import pandas as pd

config = pd.read_csv(repo_root / "transport" / "trans_config.csv")
print(f"{len(config)} rows, {len(config.columns)} columns")
config.head(10)
```

## 3 — Run the transport pipeline

The pipeline runs in two steps: `build_transport_config` (generates parameters) then `build_final_energy_demand` (calculates demand).

```python
import subprocess

result = subprocess.run(
    [sys.executable, "-m", "transport.build_transport_config"],
    capture_output=True, text=True, cwd=repo_root,
)
print(result.stdout[-2000:] if result.stdout else "(no stdout)")
if result.returncode != 0:
    print("STDERR:", result.stderr[-1000:])
```

## 4 — Inspect per-country parameter file

```python
output_dir = repo_root / "transport" / "output_data" / "country_parameters"
param_files = sorted(output_dir.glob("*.csv"))
print(f"Found {len(param_files)} country parameter files")

if param_files:
    country_params = pd.read_csv(param_files[0], index_col=0)
    print(f"\n{param_files[0].name}: {country_params.shape[0]} metrics x {country_params.shape[1]} columns")
    country_params.head(15)
```
